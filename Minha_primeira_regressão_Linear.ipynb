```python
# ===============================
# IMPORTS NECESSÁRIOS
# ===============================
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
```

```python
# ===============================
# DADOS SIMPLES (HORAS DE ESTUDO)
# ===============================
df = pd.DataFrame({
    "horas_estudo": [1, 2, 3, 4, 5, 6, 7, 8],
    "nota":         [4, 5, 6, 7, 8, 8.5, 9, 10]
})

print(" Dados utilizados:")
display(df)
```

```python
# ===============================
# DEFININDO ENTRADA (X) E SAÍDA (y)
# ===============================
X = df[["horas_estudo"]]   # variável independente
y = df["nota"]             # variável dependente

# ===============================
# DIVIDINDO TREINO E TESTE
# ===============================
X_treino, X_teste, y_treino, y_teste = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
# ===============================
# CRIANDO E TREINANDO O MODELO
# ===============================
modelo = LinearRegression()
modelo.fit(X_treino, y_treino)
```

```python
# ===============================
# FAZENDO PREVISÕES
# ===============================
pred = modelo.predict(X_teste)
```

```python
# ===============================
# AVALIANDO O ERRO
# ===============================
mae = mean_absolute_error(y_teste, pred)

print("\n Modelo treinado!")
print("Coeficiente (peso):", modelo.coef_[0])
print("Intercepto:", modelo.intercept_)
print("MAE (erro médio absoluto):", mae)
```

```python
# ===============================
# COMPARAÇÃO REAL vs PREVISTO
# ===============================
resultado = pd.DataFrame({
    "horas_estudo": X_teste["horas_estudo"].values,
    "real": y_teste.values,
    "previsto": pred
})

print("\n Comparação Real vs Previsto:")
display(resultado)
```

```python
# ===============================
# NOVA PREVISÃO
# ===============================
horas_novas = np.array([[5]])

print("\n Se estudar 5 horas, a nota prevista será:",
      modelo.predict(horas_novas)[0])
```
